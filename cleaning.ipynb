## Data Understanding


# Dataset Column Descriptions: Online Shoppers Purchasing Intention

This dataset contains user session metrics aimed at predicting online shopper purchasing intent (`Revenue`). Below is the detailed documentation for each feature:

| Column Name | Data Type | Description |
| :--- | :--- | :--- |
| **Administrative** | Integer | Number of administrative pages visited by the user during the session. |
| **Administrative_Duration** | Float | Total time spent (in seconds) on administrative pages. |
| **Informational** | Integer | Number of informational pages visited (e.g., About Us, Contact). |
| **Informational_Duration** | Float | Total time spent (in seconds) on informational pages. |
| **ProductRelated** | Integer | Number of product-related pages visited during the session. |
| **ProductRelated_Duration** | Float | Total time spent (in seconds) on product-related pages. |
| **BounceRates** | Float | Percentage of visitors who enter the site and leave after viewing only one page. |
| **ExitRates** | Float | Percentage of pageviews for a specific page that were the last in the session. |
| **PageValues** | Float | Average value of the page measured before a user completes an e-commerce transaction. |
| **SpecialDay** | Float | Closeness of the visit time to a specific holiday or special event (scaled from 0 to 1). |
| **Month** | Categorical | Month in which the user session took place. |
| **OperatingSystems** | Integer | Anonymized identifier for the user's operating system. |
| **Browser** | Integer | Anonymized identifier for the web browser used. |
| **Region** | Integer | Anonymized geographic region code of the user. |
| **TrafficType** | Integer | Anonymized code representing the source/traffic channel (e.g., Direct, Ads, Organic). |
| **VisitorType** | Categorical | Categorization of the user (`Returning_Visitor`, `New_Visitor`, `Other`). |
| **Weekend** | Boolean | Indicates whether the session occurred on a weekend (`True`/`False`). |
| **Revenue** | Boolean | **Target Class**: Indicates whether the session resulted in a purchase (`True`/`False`). |

## Data Analysis

- Question 1: What is the overall Conversion Rate and target class distribution?

```python
total_sessions = len(df)
purchases = df['Revenue'].sum()
no_purchases = total_sessions - purchases
conversion_rate = df['Revenue'].mean() * 100

print(f"Total Sessions: {total_sessions}")
print(f"Purchases (Revenue = 1): {purchases} ({conversion_rate:.2f}%)")
print(f"No Purchases (Revenue = 0): {no_purchases} ({100 - conversion_rate:.2f}%)\n")
```

```python
fig = px.pie(df, 
             names='Revenue', 
             title='Revenue Distribution (Conversion Rate)',
             labels={0: 'No Purchase', 1: 'Purchase'},
             color='Revenue')
fig.show()
```

-Business Insight: يوضح هذا أن المتجر يفقد ما يقرب من 84% من إجمالي الحركة (Traffic) الوافدة إليه، مما يستدعي دراسة سلوك الزوار في باقي الأسئلة لكشف أسباب المغادرة وتصميم استراتيجيات لتحسين نسبة الشراء.

## Machine Learning 

```python
df
```

```python
x=df.drop('Revenue', axis=1)
y=df['Revenue']
```

```python
num_col=x.select_dtypes(include=['number']).columns
obj_col=x.select_dtypes(include=['object']).columns
obj_col
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import RobustScaler
rc=RobustScaler()
num_pipeline=Pipeline(steps=[('scaler', rc)])
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
ohe=OneHotEncoder(handle_unknown='ignore', sparse_output=False)
cat_pipeline=Pipeline(steps=[('onehot', ohe)])
```

- Assign each column to its corresponding Pipeline

```python
from sklearn.compose import ColumnTransformer

preprocessing = ColumnTransformer(transformers= [ ('Num Pipeline', num_pipeline, num_col),
                                                  ('categorical Pipeline', cat_pipeline, obj_col) ])
preprocessing
```

### Selecting Best Model

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size= 0.2, random_state= 42, stratify= y)
```

```python
y_train.value_counts(normalize= True)*100
```

```python
major_class = y_train.value_counts()[0]
major_class
```

```python
minor_class=y_train.value_counts()[1]
minor_class
```

```python
scale_pos_weight = major_class / minor_class
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.model_selection import cross_validate

models = [
    ('Logistic Regression', LogisticRegression(random_state= 42, class_weight= 'balanced')),
    # ('KNN', KNeighborsClassifier()),
    ('Gaussian NB', GaussianNB()),
    # ('SVC', SVC(random_state= 42)),
    ('Decision Tree', DecisionTreeClassifier(random_state= 42, class_weight = 'balanced')),
    ('Random Forest', RandomForestClassifier(random_state= 42, class_weight = 'balanced')),
    ('XGBoost', XGBClassifier(random_state= 42, scale_pos_weight = scale_pos_weight))
]

for model in models:
    
    model_pipeline = Pipeline(steps= [('Preprocessing',preprocessing),
                                      ('Model', model[1])])
    
    cv_result = cross_validate(estimator= model_pipeline,
                                X= x_train,
                                y= y_train,
                                scoring= ['precision', 'recall'],
                                cv= 5,
                                return_train_score= True,
                                n_jobs= -1
                                )
    
    print(model[0])
    print('Train Precision :', cv_result['train_precision'].mean().round(4) * 100)
    print('Test Precision :', cv_result['test_precision'].mean().round(4) * 100)

    print('Train Recall :', cv_result['train_recall'].mean().round(4) * 100)
    print('Test Recall :', cv_result['test_recall'].mean().round(4) * 100)
    print('\n')
```

## Hyberparameter tuning

```python
rf_pipeline =  Pipeline(
    steps=[
        ('Preprocessing', preprocessing),
        ('Model', XGBClassifier(
            random_state=42,
            scale_pos_weight=scale_pos_weight
        ))
    ]
)
```

```python
param_grid = {'Model__max_depth' : [5, 6, 7],
              'Model__n_estimators' : [50, 100, 200]}
param_grid
```

```python
from sklearn.model_selection import RandomizedSearchCV

randomized_search = RandomizedSearchCV(rf_pipeline, param_grid, n_iter= 6, cv= 5, scoring= 'recall', return_train_score= True, n_jobs= -1)
```

```python
randomized_search.fit(x_train, y_train)
```

```python
randomized_search.cv_results_['mean_train_score']
```

```python
randomized_search.cv_results_['mean_test_score']
```

```python
randomized_search.best_score_
```
